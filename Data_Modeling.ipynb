```python
def create_date_dimension(start_date, end_date):
    # 1. توليد سلسلة من التواريخ بين البداية والنهاية
    dates = pd.date_range(start=start_date, end=end_date)
    
    # 2. إنشاء DataFrame
    df = pd.DataFrame(dates, columns=['FullDate'])
    
    # 3. استخراج السمات المختلفة (Attributes)
    df['DateKey'] = df['FullDate'].dt.strftime('%Y%m%d').astype(int) # مفتاح فريد بصيغة رقمية
    df['Year'] = df['FullDate'].dt.year
    df['Month'] = df['FullDate'].dt.month
    df['Day'] = df['FullDate'].dt.day
    df['Quarter'] = df['FullDate'].dt.quarter
    df['DayName'] = df['FullDate'].dt.day_name()
    df['MonthName'] = df['FullDate'].dt.month_name()
    df['DayOfWeek'] = df['FullDate'].dt.dayofweek # 0=Monday, 6=Sunday
    df['IsWeekend'] = df['DayOfWeek'].apply(lambda x: 1 if x >= 5 else 0)
    df['WeekOfYear'] = df['FullDate'].dt.isocalendar().week
    
    return df

# مثال للاستخدام: تغطية سنة 2008 (بناءً على بياناتك)
dim_date = create_date_dimension('2008-01-01', '2008-12-31')

# عرض أول 5 صفوف
print(dim_date.head())

# حفظ الملف لاستخدامه في SQL Server أو Power BI
dim_date.to_csv('Dim_Date.csv', index=False)
```
